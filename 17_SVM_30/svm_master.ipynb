### Import necessary libraries & Setup

```python
import pandas as pd
from sklearn import datasets
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.svm import SVC
from sklearn.metrics import accuracy_score

# Global list to store all experiment results dynamically
results_list = []

def log_result(exp_num, dataset_name, kernel, C, gamma, degree, accuracy):
    """Helper to append results cleanly"""
    results_list.append({
        "Experiment": exp_num,
        "Dataset": dataset_name,
        "Kernel": kernel,
        "C": C,
        "Gamma": gamma,
        "Degree": degree,
        "Accuracy": round(accuracy, 4)
    })
```

---
### Data Loading & Preprocessing Function
We create a clean helper function that loads, splits, and scales the data all at once.

```python
def prepare_data(dataset_loader):
    """Loads, splits, and scales a scikit-learn dataset."""
    data = dataset_loader()
    X_train, X_test, y_train, y_test = train_test_split(
        data.data, data.target, test_size=0.3, random_state=42
    )
    # Scale features so distance calculations work correctly
    scaler = StandardScaler()
    X_train_scaled = scaler.fit_transform(X_train)
    X_test_scaled = scaler.transform(X_test)
    
    return X_train_scaled, X_test_scaled, y_train, y_test

# Pre-process all datasets once upfront
iris_data = prepare_data(datasets.load_iris)
wine_data = prepare_data(datasets.load_wine)
cancer_data = prepare_data(datasets.load_breast_cancer)
digits_data = prepare_data(datasets.load_digits)

def run_svm(data_tuple, kernel_type, C_val, gamma_val, degree_val=3):
    """Runs SVM on pre-scaled data and returns the accuracy."""
    X_train, X_test, y_train, y_test = data_tuple
    clf = SVC(kernel=kernel_type, C=C_val, gamma=gamma_val, degree=degree_val)
    clf.fit(X_train, y_train)
    return accuracy_score(y_test, clf.predict(X_test))
```

### --- PART 1: IRIS DATASET (8 EXPERIMENTS) ---
1. Linear Kernel, C=1
2. Linear Kernel, C=10
3. RBF Kernel, C=1, gamma='scale'
4. RBF Kernel, C=1, gamma='auto'
5. Poly Kernel, C=1, degree=2
6. Poly Kernel, C=1, degree=3
7. Sigmoid Kernel, C=1
8. RBF Kernel, C=100, gamma=0.1

```python
iris_exps = [
    (1, 'linear', 1, 'scale', 3),
    (2, 'linear', 10, 'scale', 3),
    (3, 'rbf', 1, 'scale', 3),
    (4, 'rbf', 1, 'auto', 3),
    (5, 'poly', 1, 'scale', 2),
    (6, 'poly', 1, 'scale', 3),
    (7, 'sigmoid', 1, 'scale', 3),
    (8, 'rbf', 100, 0.1, 3)
]

for exp, kernel, C, gamma, deg in iris_exps:
    acc = run_svm(iris_data, kernel, C, gamma, deg)
    print(f"Exp {exp}: {acc:.4f}")
    log_result(exp, "Iris", kernel, C, gamma, deg, acc)
```
